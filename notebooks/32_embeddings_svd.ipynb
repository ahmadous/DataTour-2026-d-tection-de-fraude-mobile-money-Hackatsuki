# 32 / 33 — Embeddings SVD du graphe biparti

TruncatedSVD (k=16) sur la matrice origines × destinations (train+test, purement structurel).
En CV : +0.0005 recent2, +0.002 last — mais sur le LB : **32_emb_w50 → 0.356380** (pire que 0.357915).
Leçon : embeddings **transductifs** → la CV (validation encore "train-like") surestime le transfert
vers le vrai futur. Les doses réduites (33_tinyemb_15 → 0.357243) confirment : plus d'embeddings = pire.

```python
%load_ext autoreload
%autoreload 2
import sys
from pathlib import Path
ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np, pandas as pd
from catboost import CatBoostClassifier
from src import config as C
from src.utils import op03_mask, seed_everything, make_submission
from src.features.temporal import balance_features, recency_features
from src.features.behavioral import behavioral_features
from src.encoding import oof_target_encode_train, fit_target_map, apply_target_map, recent_target_rate
seed_everything(42)
DATA = ROOT / "data"
train = pd.read_csv(DATA / "train.csv"); test = pd.read_csv(DATA / "test.csv")
op03 = op03_mask(train).to_numpy(); y_all = train[C.TARGET].to_numpy()
te_op = op03_mask(test).to_numpy()
EPS = 1e-6; W = (5, 10, 20); SM = 30
def rowf(df):
    f = pd.DataFrame(index=df.index)
    f["amount_log1p"] = np.log1p(np.maximum(df[C.AMOUNT], 0))
    f["amount_vs_origin_before"] = df[C.AMOUNT] / (np.abs(df[C.ORIGIN_BAL_BEFORE]) + EPS)
    f["amount_vs_dest_before"] = df[C.AMOUNT] / (np.abs(df[C.DEST_BAL_BEFORE]) + EPS)
    f["origin_balance_before"] = df[C.ORIGIN_BAL_BEFORE]; f["dest_balance_before"] = df[C.DEST_BAL_BEFORE]
    return pd.concat([f, balance_features(df)], axis=1)
def bb(df, ref):
    X = rowf(df).reset_index(drop=True)
    for c in [C.ORIGIN_ACCT, C.DEST_ACCT]:
        X[f"fq_{c}"] = df[c].map(ref[c].value_counts(normalize=True)).fillna(0).values
    return pd.concat([X, behavioral_features(df, ref).reset_index(drop=True),
                      recency_features(df, ref).reset_index(drop=True),
                      recent_target_rate(df, ref, C.ORIGIN_ACCT, C.PERIOD, C.TARGET, W).reset_index(drop=True)], axis=1)
def ftr(df, ref):
    X = bb(df, ref); X["te"] = oof_target_encode_train(ref, C.ORIGIN_ACCT, C.TARGET, smoothing=SM); return X
def fap(df, ref):
    X = bb(df, ref); mp, gm = fit_target_map(ref, C.ORIGIN_ACCT, C.TARGET, smoothing=SM)
    X["te"] = apply_target_map(df, C.ORIGIN_ACCT, mp, gm); return X
def cat(seed=42):
    return CatBoostClassifier(loss_function="Logloss", eval_metric="PRAUC", depth=6,
                              learning_rate=0.05, iterations=600, random_seed=seed, verbose=False)
def rk(x):
    return np.argsort(np.argsort(x)) / (len(x) - 1)
def out(name, p_op):
    full = np.zeros(len(test)); full[te_op] = p_op
    print("écrit :", make_submission(test[C.ID], full, name))
ref0 = train.iloc[np.where(op03)[0]]; y0 = y_all[op03]
test_op = test.iloc[np.where(te_op)[0]].copy()
print("setup OK |", op03.sum(), "op03 train /", te_op.sum(), "op03 test")
```

```python
def train_champion(seed=42):
    """Champion CatBoost (config nb08) entraîné sur tout le train op_03."""
    return cat(seed).fit(ftr(ref0, ref0), y0).predict_proba(fap(test_op, ref0))[:, 1]

def pseudo_from(pch, thr_f=0.98, thr_l=0.02, seed=42):
    """Pseudo-labeling 1 cycle (cf. nb23) puis réentraînement."""
    mfm = pch > thr_f; mlm = pch < thr_l
    pse = test_op.iloc[np.where(mfm | mlm)[0]].copy()
    pse[C.TARGET] = (pch[mfm | mlm] > thr_f).astype(float)
    aug = pd.concat([ref0, pse], ignore_index=True)
    print(f"pseudo ({thr_f}/{thr_l}, seed {seed}) : {int(mfm.sum())} fraudes / {int(mlm.sum())} légitimes")
    m = cat(seed).fit(ftr(aug, aug), aug[C.TARGET].to_numpy())
    return m.predict_proba(fap(test_op, aug))[:, 1]
```

## Embeddings SVD (origines et destinations)

```python
from scipy.sparse import csr_matrix
from sklearn.decomposition import TruncatedSVD
K = 16
allo = pd.concat([train[C.ORIGIN_ACCT], test[C.ORIGIN_ACCT]]).astype("category")
alld = pd.concat([train[C.DEST_ACCT], test[C.DEST_ACCT]]).astype("category")
oidx = {a: i for i, a in enumerate(allo.cat.categories)}
didx = {a: i for i, a in enumerate(alld.cat.categories)}
ro = pd.concat([train[C.ORIGIN_ACCT], test[C.ORIGIN_ACCT]]).map(oidx).to_numpy()
cd = pd.concat([train[C.DEST_ACCT], test[C.DEST_ACCT]]).map(didx).to_numpy()
M = csr_matrix((np.ones(len(ro)), (ro, cd)), shape=(len(oidx), len(didx)))
svd = TruncatedSVD(n_components=K, random_state=42)
oemb = svd.fit_transform(M); demb = svd.components_.T
print("variance expliquée :", round(svd.explained_variance_ratio_.sum(), 3))
def emb_feats(df):
    oi = df[C.ORIGIN_ACCT].map(oidx).to_numpy(); di = df[C.DEST_ACCT].map(didx).to_numpy()
    return pd.DataFrame(np.hstack([oemb[oi], demb[di]]),
                        columns=[f"oe{i}" for i in range(K)] + [f"de{i}" for i in range(K)], index=df.index)
def bb_emb(df, ref):
    return pd.concat([bb(df, ref), emb_feats(df).reset_index(drop=True)], axis=1)
def ftr_e(df, ref):
    X = bb_emb(df, ref); X["te"] = oof_target_encode_train(ref, C.ORIGIN_ACCT, C.TARGET, smoothing=SM); return X
def fap_e(df, ref):
    X = bb_emb(df, ref); mp, gm = fit_target_map(ref, C.ORIGIN_ACCT, C.TARGET, smoothing=SM)
    X["te"] = apply_target_map(df, C.ORIGIN_ACCT, mp, gm); return X
```

## Stack embeddings : champion + pseudo, puis blends (32)

```python
pch_e = cat().fit(ftr_e(ref0, ref0), y0).predict_proba(fap_e(test_op, ref0))[:, 1]
mfm = pch_e > 0.98; mlm = pch_e < 0.02
pse = test_op.iloc[np.where(mfm | mlm)[0]].copy()
pse[C.TARGET] = (pch_e[mfm | mlm] > 0.98).astype(float)
aug = pd.concat([ref0, pse], ignore_index=True)
p1_e = cat().fit(ftr_e(aug, aug), aug[C.TARGET].to_numpy()).predict_proba(fap_e(test_op, aug))[:, 1]
# pipeline standard (sans embeddings) pour les croisements
pch = train_champion(); p1 = pseudo_from(pch)
print("corr pseudo-emb vs pseudo :", round(np.corrcoef(p1_e, p1)[0, 1], 4))
out("32_emb_w50", 0.5 * rk(pch_e) + 0.5 * rk(p1_e))
out("32_emb_x_oldpseudo", 0.5 * rk(p1_e) + 0.5 * rk(p1))
out("32_emb_triple", (rk(pch_e) + rk(p1_e) + rk(p1)) / 3)
```

## 33 — petite dose d'embeddings sur le meilleur blend

```python
base = 0.5 * rk(pch) + 0.5 * rk(p1)
for w in [10, 15, 20]:
    out(f"33_tinyemb_{w}", (1 - w / 100) * base + (w / 100) * rk(p1_e))
```
